```python
import numpy as np
```

```python
x = np.array([2.0, 3.0, 1.0])
w = np.array([0.5, -0.2, 0.8])
b = 0.1
```

```python
# z = x * w + b
z = np.dot(x, w) + b
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
y = sigmoid(z)
```

```python
print("x =", x)
print("w =", w)
print("b =", b)
print("z =", z)
print("neuron output=", y)
```

```python
# Dérivée de sigmoid
sigmoid_derivative = y * (1 - y)
print("sigmoid'(z) =", sigmoid_derivative)
```
